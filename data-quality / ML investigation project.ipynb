```python
import re

def clean_product_placeholder(row):
    text = row["Ticket Description"]
    product = row["Product Purchased"]

    # Replace product_purchased regardless of capitalization
    # and whether it has {}, [], or <> around it
    pattern = r"[\{\[\<]?\s*product_purchased\s*[\}\]\>]?"

    text = re.sub(
        pattern,
        product,
        text,
        flags=re.IGNORECASE
    )

    return text


df["Clean Description"] = df.apply(
    clean_product_placeholder,
    axis=1
)
```

```python
remaining = df[
    df["Clean Description"].str.contains(
        r"product_purchased",
        case=False,
        na=False
    )
]

print("Remaining rows:", len(remaining))
```

```python
X_text = df["text"]
y = df["Ticket Type"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
vectorizer = TfidfVectorizer()
```

```python
X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)
```

```python
print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(
    max_iter=1000
)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
print("Actual distribution:")
print(y_test.value_counts())

print("\nPredicted distribution:")
print(pd.Series(y_pred).value_counts())
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=model.classes_
)

disp.plot(xticks_rotation=45)
plt.title("Confusion Matrix - Version 1")
plt.show()
```
